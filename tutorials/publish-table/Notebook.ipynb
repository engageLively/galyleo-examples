# Notebook
This notebook reads the ufos.sdml file from disk, and then publishes it to the Galyleo Service.
Step 1 is import the libraries.

```python
import json
import requests
import os
```

Step 2 is to define a function which publishes a table.  This is a  convenience, but it's worthwhile
when we'll be doing it multiple times.  The URL for publication is `<HUB_URL>/services/publish_data`, where `HUB_URL` is the URL of the JupyterHub.  The API token pre-loaded into `JUPYTERHUB_API_TOKEN` is required for authentication; and then the body of the request is the table.

```python
JUPYTER_HUB_ROOT_URL = 'https://galyleokct-free.engagelively.com'
publish_url = f'{JUPYTER_HUB_ROOT_URL}/services/galyleo/publish_data'
api_token = os.getenv('JUPYTERHUB_API_TOKEN', '')
headers = {"Authorization": f"token {api_token}"}
def publish_table(table, name):
    if not name.endswith('.sdml'):
        name = name + '.sdml'
    payload = {
        "table": table,
        "name": name
    }
    headers = {'Authorization': f'token {api_token}'}
    return requests.post(publish_url, headers = headers, json = payload)
    
```

Finally, use the publish method defined above; open each file, and pass its name and body to `publish_table` as arguments.

```python
for name in ['ufos_cy.sdml', 'ufos_cym.sdml']:
    with open(name, 'r') as f:
        table_spec = json.load(f)
        publish_table(table_spec, name)
```
